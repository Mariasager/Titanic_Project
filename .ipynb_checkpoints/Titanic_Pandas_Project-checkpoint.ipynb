## Basic Statistics

```python
import seaborn as sns
import pandas as pd
df= sns.load_dataset("titanic")

#What is the average age of passengers?
df["age"].mean()
#What is the average fare paid?
df["fare"].mean()
#How many passengers were male vs female?
df["sex"].value_counts()
#How many passengers survived vs not survived
df["survived"].value_counts()
```

## Grouping & Comparison

```python
import seaborn as sns
import pandas as pd
df= sns.load_dataset("titanic")

#What was the average age of survivors vs non-survivors?
df.groupby("survived")["age"].mean()
#What was the survival rate by gender?
df.groupby("sex")["survived"].mean()
#What was the survival rate by passenger class (1st, 2nd, 3rd)?
df.groupby("pclass")["survived"].mean()
#What was the average fare paid by each class?
df.groupby("pclass")["fare"].mean()
#Which class had the highest survival rate?
df.groupby("pclass")["survived"].mean().idxmax()
#Create a new column family_size = sibsp + parch + 1.
df["family_size"]=df["sibsp"]+df["parch"]+1
df.head()
#On average, did people with larger families have higher or lower survival rate?
df.groupby("family_size")["survived"].mean()

#Save cleaned dataset
df.to_csv("titanic_cleaned.csv", index=False)
```
